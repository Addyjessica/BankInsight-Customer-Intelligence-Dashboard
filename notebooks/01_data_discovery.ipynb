```python
transaction_summary = (
    transactions_df
    .groupby("customer_id")
    .agg(
        calculated_tx_count=("amount", "count"),
        calculated_avg_tx_value=("amount", "mean"),
        calculated_total_tx_volume=("amount", "sum")
    )
    .reset_index()
)

transaction_summary.head()
```

```python
validation = customer_df.merge(
    transaction_summary,
    on="customer_id",
    how="left"
)

validation.head()
```

```python
(validation["tx_count"] == validation["calculated_tx_count"]).value_counts()
```

```python
(validation["total_tx_volume"] == validation["calculated_total_tx_volume"]).value_counts()
```

# Data Discovery Report
## Objective
The objective of this phase was to just try and understand the structure, quality, and the relationships within the customer and transactions datasets before performing any data prepation or analysis
## Dataset Overview
The project consists of two datasets:
* Customer Dataset: 48723 records and 54 variables containing customer demographics, banking product, ownership, digital engagement, customer satisfaction, and precomputed customer metrics
* Transaction Dataset: 3,159,157 transaction records containing transaction dates, amounts, and transaction types.
The two datasets are linked through the **customer_id** field in a one-to-many relationship, where one customer may have multiple transactions.
# Data Quality Assessment
## Customer Dataset
The customer dataset demonstrated excellent overall quality.
Three variables contained missing values:

| Variable	| Missing (%)	| Initial Interpretation |
| ----------| --------------| ------------------------|
| credit_utilization_ratio|	37.48%	|Likely not applicable for customers without a credit card|
| feature_requests	|33.07%	|Likely indicates customers who have not submitted feature | requests
|complaint_topics	|49.97%|	Likely indicates customers who have not lodged complaints|

All remaining variables contained complete observations
## Transaction Dataset
The transaction dataset contained no missing values across any of its variables.
# Data Validation
Several validation checks were performed to assess the integrity of the data:
* Relashionship Integrity - Every transaction record was successfully matched to a customer in the customer dataset.
* Metric Validation - The customer-level summary metrics (tx_count, avg_tx_value, and total_tx_volume) were recalculated from the transaction dataset and matched the values provided in the customer dataset.
# Initial Observations
The following observations were made:
* The dataset follows a relational structure commonly found in banking and financial systems.
* The transaction dataset provides sufficient detail to derive customer-level behavioral metrics.
* Several variables, including Customer Lifetime Value (CLV) and Churn Probability, appear to be derived metrics rather than raw observations.
* Some transaction summary variables appear duplicated under different names and will be reviewed during data preparation.
* Missing values appear to follow logical business rules rather than indicating poor data quality.

# Decisions for the data cleaning phase
Based on the discovery phase, the following actions will be performed during data preparation:
- Convert all date fields to datetime format.
- Validate and remove duplicate records if necessary.
- Preserve missing values that represent "not applicable" rather than data errors.
- Replace missing values in feature requests and complaint topics with meaningful business categories.
- Review duplicated summary variables and determine whether they should be retained or removed.
- Optimize data types where appropriate.
# Conclusions
The datasets demonstrate a high level of completeness and consistency, making them suitable for downstream analysis. Validation checks confirmed the integrity of the relationship between customer and transaction data as well as the accuracy of customer-level transaction summary metrics.
