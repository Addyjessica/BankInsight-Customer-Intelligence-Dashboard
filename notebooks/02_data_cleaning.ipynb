# Customer Intelligence Platform for a Digital Bank

## Phase 2: Data Preparation
## Business Objective

The objective of this phase is to prepare the customer and transaction datasets for analysis while preserving the integrity and business meaning of the data.

This phase includes correcting data types, handling missing values, validating business rules, optimizing data structures, and producing clean datasets that will support reliable exploratory analysis and dashboard development.

```python
import pandas as pd
import numpy as np
customer_df = pd.read_csv('../data/customer_data.csv')
transactions_df = pd.read_csv('../data/transactions_data.csv')
```

```python
customer_df.head()
```

```python
transactions_df.head()
```

```python
customer_date_columns = [
    "first_tx",
    "last_tx",
    "last_survey_date",
    "last_transaction_date",
    "first_transaction_date"
]

for column in customer_date_columns:
    customer_df[column] = pd.to_datetime(customer_df[column])

transactions_df["date"] = pd.to_datetime(transactions_df["date"])
```

```python
customer_df["feature_requests"] = (
    customer_df["feature_requests"]
    .fillna("No Request")
)
```

```python
customer_df["complaint_topics"] = (
    customer_df["complaint_topics"]
    .fillna("No Complaint")
)
```

```python
customer_df.groupby("credit_card")[
    "credit_utilization_ratio"
].apply(lambda x: x.isna().sum())
```

```python
transactions_df["amount"].describe()
```

```python
(transactions_df["amount"] <= 0).sum()
```

```python
categorical_columns = [
    "gender",
    "location",
    "income_bracket",
    "occupation",
    "education_level",
    "marital_status",
    "acquisition_channel",
    "customer_segment",
    "feedback_sentiment",
    "clv_segment",
    "preferred_transaction_type",
    "type"
]
```

```python
customer_categorical_columns = [
    "gender",
    "location",
    "income_bracket",
    "occupation",
    "education_level",
    "marital_status",
    "acquisition_channel",
    "customer_segment",
    "feedback_sentiment",
    "clv_segment",
    "preferred_transaction_type"
]

for column in customer_categorical_columns:
    customer_df[column] = customer_df[column].astype("category")

transactions_df["type"] = transactions_df["type"].astype("category")
```

```python
customer_df.info()

transactions_df.info()
```

```python
customer_df.to_csv(
    "../data/customers_clean.csv",
    index=False
)

transactions_df.to_csv(
    "../data/transactions_clean.csv",
    index=False
)
```

## Data Preparation Summary

The customer and transaction datasets were successfully prepared for analysis.

Key preparation activities included:

- Converted all date fields to datetime format.
- Verified and handled duplicate records.
- Replaced missing values in business-related categorical fields (`feature_requests` and `complaint_topics`) with meaningful labels.
- Preserved missing values in `credit_utilization_ratio` pending business-rule validation.
- Optimized selected categorical variables to improve memory efficiency.
- Validated transaction amount values for potential anomalies.
- Saved clean datasets for use in subsequent analysis.

The resulting datasets are now ready for feature engineering and exploratory data analysis.
